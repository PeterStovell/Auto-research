```python
import sys
import os
import pandas as pd

# Add src directory to path
sys.path.append(os.path.abspath('../src'))

from utils import conf_read
from columns import Columns
from features import add_features
```

```python
# Load configuration and columns helper
cfg = conf_read('../src/dataloaders.yaml')
columns = Columns(cfg.columns)

# Load the raw dataframe as done in dataloaders.py
print(f'Loading {cfg.input_path}')
df = pd.read_parquet(cfg.input_path, columns=columns.load_list()).dropna()

if "query" in cfg:
    print("Applying query:", cfg.query)
    df = df.query(cfg.query)

columns.add_auxiliary(df)
new_features = add_features(df, columns)

print(f'Shape: {df.shape}')
df.head()
```
